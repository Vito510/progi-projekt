```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.firefox.service import Service
from selenium.webdriver.common.action_chains import ActionChains

import time
import json
import random

# service = Service("/snap/bin/geckodriver")
driver = webdriver.Chrome()
actions = ActionChains(driver)
```

**TEST Kreacija rute**

- Ucitava to podrucje

- Brise zadnjih 5

```python
driver.get("http://localhost:5173/map")

zoom_in_button = driver.find_element(By.CLASS_NAME, "leaflet-control-zoom-in")
for i in range(5):
    time.sleep(0.5)
    zoom_in_button.click()

load_start = time.time()

odaberi_podrucje_button = driver.find_element(By.CLASS_NAME, "coord-button")
odaberi_podrucje_button.click()

while len(driver.find_elements(By.TAG_NAME, "button")) < 5:
    time.sleep(0.1)

print("Reljef ucitan u", time.time()-load_start)

naziv, x_range, y_range = driver.find_elements(By.TAG_NAME, "input")
nova_tocka = driver.find_elements(By.TAG_NAME, "button")[-1]
for i in range(10):

    actions.click_and_hold(x_range).move_by_offset(random.randint(-20,20), 0).release().perform()
    actions.click_and_hold(y_range).move_by_offset(random.randint(-20,20), 0).release().perform()
    nova_tocka.click()

for i in range(5):
    trash_button = driver.find_element(By.CLASS_NAME, "square")
    trash_button.click()

naziv.clear()
naziv.send_keys(str(random.randint(100000,999999)))

spremi_button = driver.find_elements(By.TAG_NAME, "button")[1]
spremi_button.click()
```
